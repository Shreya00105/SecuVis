```python
# visualization.ipynb

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
import joblib

FEATURE_PATH = "../data/cleaned_logs/features.csv"
MODEL_PATH = "../models/if_model.pkl"
SCALER_PATH = "../models/scaler.pkl"

df = pd.read_csv(FEATURE_PATH)

model = joblib.load(MODEL_PATH)
scaler = joblib.load(SCALER_PATH)

X_scaled = scaler.transform(df)
preds = model.predict(X_scaled)

df['anomaly'] = preds

# Plot: hourly log activity
plt.figure(figsize=(10,5))
df['hour'].value_counts().sort_index().plot(kind='bar')
plt.title("Logs by Hour")
plt.xlabel("Hour")
plt.ylabel("Count")
plt.show()

# Plot: anomaly distribution
plt.figure(figsize=(6,4))
df['anomaly'].value_counts().plot(kind='bar')
plt.title("Anomaly vs Normal Count")
plt.xticks([0,1], ['Normal','Anomaly'], rotation=0)
plt.ylabel("Count")
plt.show()

```
