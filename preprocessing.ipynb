```python
# preprocessing.ipynb

import os
import pandas as pd
from utils.log_parser import parse_auth_log

RAW_LOG_DIR = "../data/raw_logs/"
CLEAN_DIR = "../data/cleaned_logs/"
os.makedirs(CLEAN_DIR, exist_ok=True)

all_docs = []

for file in os.listdir(RAW_LOG_DIR):
    if file.endswith(".log"):
        path = os.path.join(RAW_LOG_DIR, file)
        print("Processing:", file)
        parsed = parse_auth_log(path)
        all_docs.extend(parsed)

df = pd.DataFrame(all_docs)
print("Parsed Rows:", len(df))

# Basic cleaning
df.drop_duplicates(inplace=True)
df.dropna(subset=["raw_message"], inplace=True)

df.to_csv(os.path.join(CLEAN_DIR, "cleaned_logs.csv"), index=False)
df.head()
```
